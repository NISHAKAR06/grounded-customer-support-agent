# 05 - Baseline 1: Majority Class Classifier

## Overview
Establishes the trivial statistical floor for intent classification by unconditionally predicting the most frequent class in the training split (`GENERAL_INQUIRY`).

### Evaluation Datasets:
1. **Held-out Test Split** (484 real conversations from `data/splits/test.jsonl`)
2. **Golden Evaluation Set** (200 hand-verified conversations from `data/golden/golden_set.jsonl`)

```python
import json
from pathlib import Path
import joblib
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report

model_path = Path('../models/intent_classifier/majority_baseline.joblib')
test_path = Path('../data/splits/test.jsonl')
golden_path = Path('../data/golden/golden_set.jsonl')

model = joblib.load(model_path)
print(f"Majority Class: {model.majority_class} ({model.majority_ratio * 100:.2f}% of training split)")
```

## 1. Evaluation on Held-Out Test Split (484 Samples)

```python
test_data = [json.loads(l) for l in open(test_path, encoding='utf-8') if l.strip()]
X_test = [r['first_inquiry'] for r in test_data]
y_test = [r['intent_code'] for r in test_data]

preds_test = model.predict(X_test)
acc_test = accuracy_score(y_test, preds_test)
print(f"Held-Out Test Accuracy: {acc_test * 100:.2f}%")
print(classification_report(y_test, preds_test, zero_division=0))
```

## 2. Evaluation on Hand-Verified Golden Set (200 Samples)
On the balanced golden set, the trivial majority baseline drops precipitously, proving the necessity of learned semantic representations.

```python
golden_data = [json.loads(l) for l in open(golden_path, encoding='utf-8') if l.strip()]
X_gold = [r['customer_message'] for r in golden_data]
y_gold = [r['gold_intent_code'] for r in golden_data]

preds_gold = model.predict(X_gold)
acc_gold = accuracy_score(y_gold, preds_gold)
print(f"Golden Set Accuracy: {acc_gold * 100:.2f}%")
print(classification_report(y_gold, preds_gold, zero_division=0))
```
